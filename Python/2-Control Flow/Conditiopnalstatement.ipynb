```python
### If Statement
# Evaluates a condition and executes a block of code within it, if the condition is True

age = 20

if age >= 18:
    print(f"You are allowed to vote in the elections")
```

```python
age >= 18
```

```python
## else
## The else statement executes a block of code if the condition in the if statement is False.

age = 16
if age >= 18:
    print("You are eligible for voting")
else:
    print("You are a minor")
```

```python
# elif
## The elif statement allows you to check multiple conditions. It stands for "else if"

age = 20

if age < 13:
    print("You are a child")
elif age < 18:
    print("You are a teenager")
else:
    print("You are an adult")
```

```python
## elif Example

age = 17

if age < 13:
    print("You are a child")
elif age < 18:
    print("You are a teenager")
else:
    print("You are an adult")
```

#### Nested Conditional Statement

```python
## Nested Conditional STatement
## You can place one or more if, elif, or else statements inside another if, elif, or else statement to create nested conditional statement

## Number if its even, odd, or negative

num = int(input("Enter a number"))

if num > 0:
    print("The number is positive")
    if num % 2 == 0:
        print("The number is even")
    else:
        print("The number is odd")

else:
    print("The number is zero or negative")

# used number is 14
```

##### Practical Example

```python
## Determine if a year is a leap year using nested conditional statement

year = int(input("Enter the year"))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")

else:
    print(f"{year} is not a leap year")
```

```python
# More practical
# Simple calculator Program

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
operation = input("Enter operation (+, -, *, /): ")

# Perform the request operation
if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    result = num1 / num2
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error! Division by zero."
else:
    result = "Invalid operation."

print("Result:", result)
```

```python
## Practical
## Determine the ticket price based on age whether the person is a student
## Ticket pricing based on age and student status

# Take user input
age = int(input("Enter your age: "))
is_student = input("Are you a student? (yes/no): ").lower()

## Determine the ticket price
if age < 5:
    price = "Free"
elif age <= 12:
    price = "$10"
elif age <= 17:
    if is_student == 'yes':
        price = "$12"
    else:
        price = "$15"
elif age <= 64:
    if is_student == 'yes':
        price = "$18"
    else:
        price = "$25"
else:
    price = "$20"
print("Ticket Price:", price)

## Age 14 and is a student
```
