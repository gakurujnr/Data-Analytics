##### Introduction to Functions

 -  Function is a block of code that performs a specific task.

 - Functions help in organizing code, reusing code and improving readability.

```python
## Basic syntax of a function in Python:
def function_name(parameters):
    """Docstring"""  ## States what the function does. It is optional but recommended.
    # Function Body
    return expression  ## Optional. If not specified, the function will return None by default.
```

```python
## Why functions?
##
num=24
if num % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
## Using functions

def even_or_odd(num):
    """"This function finds even or odd numbers"""
    if num % 2 == 0:
        print("The number is even")
    else:
        print("The number is odd")
```

```python
## Call this function
even_or_odd(24)  ## Output: The number is even
```

```python
##
even_or_odd(21)
```

```python
## Function with multiple parameters
def add(a,b):
    """This function adds two numbers"""
    # c= a + b
    # return c
    return a+b

result = add(5,6)
print(result)
```

```python
## Default Parameters

def greet(name):
    print(f"Hello {name} welcome to the world")

greet("Gacheri")
```

```python
## With Default Values

## Default Parameters

def greet(name="Guest"):
    print(f"Hello {name} welcome to the world")

greet("Gacheri")
# greet(name="Gacheri")
```

```python
## Variable Length Arguments
## Position and Keyword Arguments

def print_numbers(*args):
    """This function prints all the numbers passed to it"""
    for num in args:
        print(num)
```

```python
## Calling the function

print_numbers(1, 2, 3, 4, 5, "Kasongo")
```

```python
## Positional Arguments
def print_numbers(*args):
    """This function prints all the numbers passed to it"""
    for number in args:
        print(number)
```

```python
##
print_numbers(1, 2, 3, 4, 5, "Kasongo")
```

```python
## Key Words Arguments

def print_details(**kwargs):
    """This function prints the details passed to it"""
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_details(name="Gacheri", age="0.1", country="Kenya")
```

```python
## Combining Positional and Keyword Arguments

def print_details(*args, **kwargs):
    for val in args:
        print(f"Positional argument: {val}")

    for key, value in kwargs.items():
            print(f"{key}: {value}")
```

```python
print_details(1,2,3,4,5, name="Gacheri", age="0.1", country="Kenya")
```

```python
## Return statements
# Our function can return a value using the return statement. If no return statement is specified, the function will return None by default.

def multiply(a, b):
    """This function multiplies two numbers"""
    return a * b

result = multiply(5, 3)
print(result)  
```

```python
## Returning Multiple Parameters

def multiply(a, b):
    """This function multiplies two numbers"""
    return a * b, a

result = multiply(5, 3)
print(result)  
```
