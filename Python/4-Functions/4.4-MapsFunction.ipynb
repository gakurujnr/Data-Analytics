### The Map() Function in Python

- The map() function applies a given function to all items in an input list(or any other iterable) and returns a map object(an iterator).

- This is particularly useful for transforming data in a list comprehensively.

```python
##

def square(x):
    return x*x

square(10)
```

```python
##

numbers = [1,2,3,4,5,6,7,8]  # Iterables

list(map(square, numbers)) 
```

```python
## Lambda function with map

numbers = [1,2,3,4,5,6,7,8]
list(map(lambda x:x*x, numbers))
```

```python
## Map Multiple Iterables

numbers1 = [1,2,3]
numbers2 = [4,5,6]

added_numbers = list(map(lambda z,y:z+y, numbers1, numbers2))
print(added_numbers)
```

```python
## Map() to convert a list of string to integers

str_numbers = ['1','2','3','4','5']
int_numbers = list(map(int, str_numbers))

print(int_numbers)

```

```python
## In-built function with the help of map

words = ['apple', 'banana', 'cherry']
upper_words = list(map(str.upper,words))

# map() → applies a function to each item in an iterable

# str.upper → the function being applied (turns a string into uppercase)

# list(...) → converts the map object into a list

print(upper_words)
```

```python
## 
def get_name(person):
    return person['name']

people = [
    {'name':'AG', 'age': 29},
    {'name': 'Persona', 'age': 40}

]
list(map(get_name, people))
```

#### Conclusion

- The map() function is a powerful tool for transforming iterable data.

- It works with regular functions, lambda functions, and multiple iterables.

- It provides a versatile approach to data processing in Python.

- Using map() helps you write more efficient and readable code.
