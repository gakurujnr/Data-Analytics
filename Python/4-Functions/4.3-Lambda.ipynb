### Lambda Functions in Python

 - Lambda functions are a small anonymous functions defined using the lambda keyword.

 - They can have any number of arguments but only one expression.

 - They are commonly used for short operation or as arguments to higher-order functions.

```python
## Syntax
lambda arguments: expression
```

```python
## Create a function

def addition(x, y):
    return x + y
```

```python
addition(4, 5)
```

```python
## Lambda Function
addition  = lambda x,y: x +y
print(addition)
print(addition(4, 5))
```

```python
##
def even(num):
    if num %2 == 0:
        return True
    
even(22)
```

```python
##
even1 = lambda num: num % 2 == 0

# even1(10)
even1(9)
```

```python
## Multiple Parameters

def addition(x,y,z):
    return x+y+z

addition(12,11,45)
```

```python
## Lambda

addition1 = lambda x,y,z: x+y+z
addition(22,45,32)
```

##### Example of Map in Lambda

```python
## Map () - Applies a function to all items in a list.

number = [1,2,3,4,5,6]

def square(number):
    return number**2

square(2)
```

```python
## Using Map
square1 = list(map(lambda x: x**2, number))

print(square1)
```
