### Assignment 3: Data Cleaning

1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.


```python
##
data = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)), columns=['A', 'B', 'C'])
print("Original Data:\n", data)

## Introduced NaN values
data.iloc[0, 1] = np.nan
data.iloc[2, 2] = np.nan
data.iloc[4, 0] = np.nan
print("DataFrame with NaN Values:\n", data)

## Fill the NaN values with Mean
data = data.fillna(data.mean(), inplace=True)
print("DataFrame with NaN values Filled:\n", data)
```

### Assignment 4: Data Aggregation

1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.


```python
##
dt = pd.DataFrame({'Category': np.random.choice(['A', 'B', 'C'], size=10), 'Value': np.random.randint(1, 100, size=10)})
print("Original DataFrame:\n", dt)

# Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category
grouped = dt.groupby('Category')['Value'].agg(['sum', 'mean'])
# df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("Grouped by DataFrame:\n", grouped)
```

2. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
# Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'
dt = pd.DataFrame({'Product': np.random.choice(['Prod1', 'Prod2', 'Prod3'], size=10), 'Category': np.random.choice(['A', 'B', 'C'], size=10), 'Sales': np.random.randint(1, 100, size=10)})
print("Original DataFrame:\n", dt)

## Group the DataFrame by 'Category' and compute the total sales for each Category
dt = dt.groupby('Category')['Sales'].sum()
print("Computed DataFrame:\n", dt)
```

1. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.


```python
##
dt1 = pd.DataFrame({'Key': ['A', 'B', 'C', 'D'], 'Value1': np.random.randint(1, 100, size=4)})
dt2 = pd.DataFrame({'Key': ['A', 'B', 'C', 'E'], 'Value2': np.random.randint(1, 100, size=4)})
print("DataFrame 1:\n", dt1)
print("DataFrame 2:\n", dt2)

# Merge the DataFrames with common column
dt = pd.merge(dt1, dt2, on='Key')
print("Merged DataFrames:\n", dt)
```

### Assignment 6: Time Series Analysis

1. Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.

```python
## Pandas DataFrame with Datetime index and one column filled with random integers
dt_rng = pd.date_range(start='2022-01-01', end='2022-12-31', freq='D')
df = pd.DataFrame(dt_rng, columns=['date'])
df['data'] = np.random.randint(0, 100, size=(len(dt_rng)))
df.set_index('date', inplace=True)
print("Original DataFrame:\n", df)

# Resample the DataFrame to compute the monthly mean of the values
mon = df.resample('ME').mean()
print("Monthly Mean DataFrame:\n", mon)
```

2. Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.

```python
# Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31'
import pandas as pd
import numpy as np

date_rng = pd.date_range(start='2021-01-01', end='2021-12-31', freq='D')
# Create a sequence of daily dates from Jan 1, 2022 to Dec 31, 2022. This is a DatetimeIndex.

df = pd.DataFrame(date_rng, columns=['date'])
# Make a DataFrame with a single column named "date" containing those daily timestamps.

df['data'] = np.random.randint(0, 100, size=(len(date_rng)))
# Add a column called "data" filled with random integers between 0 and 99.
# size=(len(dt_rng)) makes sure there is one random number for each date.

df.set_index('date', inplace=True)
print("Original DataFrame:")
print(df)

# Compute the rolling mean with a window of 7 days
rolling_mean = df.rolling(window=7).mean()


print("Rolling mean DataFrame:")
print(rolling_mean)
```
