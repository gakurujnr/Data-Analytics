2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

```python
## 0 = rows because row index starts at 0
## 1 = columns because column index is the second dimension

array = np.arange(1, 17).reshape(4, 4)
print("Array")
print(array)

## Perform row-wise and column-wise sum
rw_sum = np.sum(array, axis=0)  # axis=0 affects the Rows
clm_sum = np.sum(array, axis=1) # axis = 1 affects columns

print("Row wise sum is:\n", rw_sum)
print("Column wise sum is:\n", clm_sum)
```

### Assignment 4: Statistical Operations

2. Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

```python
##
arr = np.arange(1, 10).reshape(3, 3)
print("Original array is:\n", arr)

# Normalization
mean = np.mean(arr)
std_dev = np.std(arr)
norm = (arr - mean) / std_dev
print("Normalization:\n", norm)
```

### Assignment 6: Linear Algebra

1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.


```python
#
matrix = np.random.randint(1, 10, size=(3,3))
print("Matrix:\n", matrix)

## Computing Determinant
# determinant = 0 means matrix is not invertible
determinant = np.linalg.det(matrix)
print("Determinant:\n", determinant)

## Computing Inverse
# only works if determinant is not zero
inverse = np.linalg.inv(matrix)
print("Inverse:\n", inverse)

## Computing Eigenvalues
# eigenvalues show how the matrix scales space
eigen = np.linalg.eigvals(matrix)
print("Eigenvalues:\n", eigen)
```

### Assignment 9: Structured Arrays

1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.


```python
##
# It defines a NumPy structured data type with three fields:
#  a string name (max 10 chars), a 32‑bit integer age, and a 32‑bit float weight.
data_type = [('name', 'U10'), ('age', 'i4'), ('weight', 'f4')]
data = np.array([('Alice', 25, 55.5), ('Bob', 30, 85.3), ('Charlie', 20, 65.2)], dtype=data_type)
print("Original Array")
print(data)

## Sort the Array by age
sorted_data = np.sort(data, order="age")
print("Sorted Array")
print(sorted_data)
```

### Assignment 10: Masked Arrays

1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.


```python
## 
import numpy.ma as ma

# Create a masked array of shape (4, 4) with random integers
array = np.random.randint(1, 21, size=(4, 4))
masked_array = ma.masked_greater(array, 10)
print("Original array:")
print(array)
print("Masked array:")
print(masked_array)

# Compute the sum of the unmasked elements
sum_unmasked = masked_array.sum()
print("Sum of unmasked elements:", sum_unmasked)
```

2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

```python
# Create a masked array of shape (3, 3) with random integers
array = np.random.randint(1, 21, size=(3, 3))
masked_array = ma.masked_array(array, mask=np.eye(3, dtype=bool))
print("Original array:")
print(array)
print("Masked array:")
print(masked_array)

# Replace the masked elements with the mean of the unmasked elements
mean_unmasked = masked_array.mean()
masked_array = masked_array.filled(mean_unmasked)
print("Modified masked array:")
print(masked_array)
```
