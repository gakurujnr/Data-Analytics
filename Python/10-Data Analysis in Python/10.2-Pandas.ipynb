#### Pandas-DataFrame And Series

- Pandas is a powerful Python library used for data manipulation, data cleaning, and data analysis.

- It provides two main data structures:

- Series — a one‑dimensional, array‑like structure.

- DataFrame — a two‑dimensional, table‑like structure with rows and columns.

- A DataFrame can hold different types of data and can be resized.

- Both structures are essential for working with datasets in Python.

#### Data Manipulation with DataFrames

```python
## Remove a column
df.drop('Salary', axis=1, inplace=True) ## axis=1 is temporary, while including inplace=True, deletes permanently
df
```
