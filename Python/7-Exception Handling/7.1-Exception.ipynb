#### Understanding Exceptions

- Exception handling in Python lets you deal with errors gracefully.

- It allows your program to continue running instead of crashing.

- You use try, except, else, and finally blocks to manage errors.


#### What are Exceptions?

- Exceptions are events that interrupt the normal flow of a program.

- They occur when Python encounters an error during execution.

- Common exceptions include:

    - ZeroDivisionError — Dividing by zero.

    - FileNotFoundError — file does not exist.

    - ValueError — invalid value given.

    - TypeError — wrong data type used.

```python
## Exception try, except block
## try

try:
    a=b
except:
    print("The variable has not been assigned")
```

```python
## This is a name error, class that handles NameError
try:
    a=b
except NameError as ex:   # This block catches the NameError instead of letting the program crash.
    print(ex)
```

##### Explanation 

- except ValueError:
    - runs when the user enters something that cannot be converted to an integer

- except ZeroDivisionError:
    - runs when the user enters 0 and causes division by zero

- except Exception as ex:
    - catches any other unexpected error and prints the error message


```python
## ValueError, ZeroDivisionError and Exception

try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:  # runs when the user enters something that cannot be converted to an integer
    print("This is not a valid number")
except ZeroDivisionError: # runs when the user enters 0 and causes division by zero
    print("Enter denominator greater than zero")
except Exception as ex: 
    print(ex) # catches any other unexpected error and prints the error message
```

```python
## Try, except, else block

try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:
    print("Enter a valid number")
except ZeroDivisionError:
    print("You can't divide by zero")
except Exception as ex:
    print(ex)
else:  ## This is when there is no error and operation has to be excecuted
    print(f"The result is: {result}")

## Input 23
```

```python
## Try, except, else and finally

try:
    num = int(input("Enter a number"))
    result = 10/num
except ValueError:
    print("Enter a valid number")
except ZeroDivisionError:
    print("You can't divide by zero!")
except Exception as ex:
    print(ex)
else:  ## This is when there is no error and operation has to be excecuted.
    print(f"The result is: {result}")
finally: ## Has to be excecuted whether or not there is error
    print("Excecution is Complete.")

## Finally assist in DB to close the connection
```

```python
## File Handling and Exception Handling

try:
    file = open('example1.txt','r')
    content = file.read()
    print(content)
    ## Tries to open the file, read its content, and print them

except FileNotFoundError:  # Handle the case where file does not exist
    print("File does not exist")
    ## Runs if the file does not exist or the name is wrong

finally:
    ## This block ALWAYS runs - whether there was an error or not
    if 'file' in locals() and not file.closed():
        # check if the file object was created and is still open
        file.close()
        print('file close')
        # Closes the file to free system resources
```
