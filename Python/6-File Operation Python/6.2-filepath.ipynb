#### Joining Paths

```python
## Joining Paths

dir_name = "folder"
# name of the directory you want to work with

file_name = "file.txt"
# name of the file inside that directory

full_path = os.path.join(dir_name, file_name)
# safely combines the folder and file into one correct path (handles slashes automatically)

print(full_path)
# displays the full path (folder/file.txt)
```

```python
##

dir_name = "folder"
# the name of the directory you want to work with

file_name = "file.txt"
# the file you want to access inside that directory

full_path = os.path.join(os.getcwd(), dir_name, file_name)
# builds a full path by combining:
# - the current working directory
# - the folder name
# - the file name
# ensures the path is correct on any operating system

print(full_path)
# prints the complete file path
```

```python
## Checking if the path is a file or Directory

import os
# import os loads Python's OS module
# it lets you work with the operating system (paths, files, folders)


path = 'example.txt'
# the path you want to check (could be a file or a folder)

if os.path.isfile(path):
    # checks if the path points to an existing file
    print(f"The path '{path}' is a file")

elif os.path.isdir(path):
    # checks if the path points to an existing directory
    print(f"The path '{path}' is a directory")

else:
    # runs if the path is neither a file nor a directory (doesn't exist)
    print(f"The path '{path}' is neither a file nor a directory")
```

```python
## Absolute Pat - Is a complete path
relative_path = 'example.txt'
absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
