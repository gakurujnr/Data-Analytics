```python
## Data Serialization

## Working with JSON
# json.dumps() converts a Python object into a JSON string
# json.loads() converts a JSON string back into a Python object


import json

data = {'name':'Cherry', 'age':30}
json_str = json.dumps(data)
print(type(json_str))
print(json_str)

parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

### Creating CSV File

```python
## csv

import csv

# Opens (or creates) a file named example.csv.
# mode='w' means write mode — it will overwrite the file.
# newline='' prevents extra blank lines in the CSV.
# as file gives you a file object called file.

with open('example.csv', mode='w', newline='') as file:
    writer= csv.writer(file)   # Creates a CSV writer object that will write rows into the file.
    writer.writerow(['name','age']) # writing the rows
    writer.writerow(['Cherry', 30])

with open('example.csv', mode='r') as file:  # Opens the same CSV file again, but this time in read mode (mode='r').
    reader = csv.reader(file) # Creates a CSV reader object that will read rows from the file.
    for row in reader: 
        print(row)

```

### Datetime

```python
## Datetime

from datetime import datetime,timedelta

now = datetime.now()
print(now)

## One day back
yesterday = now-timedelta(days=1)

print(yesterday)
```

```python
## Regular Expression
# import re brings in Python's regular expression module
# used for finding patterns in text (searching, matching, replacing)


import re

pattern = r'\d+'  ## '\d+' checking for digit
text='There are 123 apples'

## Match whether the pattern is there or not
match = re.search(pattern,text)
print(match.group())
```
