### Variables

Variables are fundamental elements in programming used to store data that can be referenced and manipulated in a program. In Python, variables are created when you assign a value to them, and they do not need explicit declaration to reserve memory space. The declaration happens automatically when you assign a value to a variable.

```python
# Creating a variable
a=100
```

```python
# Declaring and Assigning a variable
age = 30
height = 5.9
name = "Gakuru"
is_student = False

# printing the variables
print("Age :", age)
print("Height :", height)
print("Name :", name)
```

```python
# Naming Convections
## Variables ames should be descriptive
## They must start with a letter or an '_' and contain letter, numbers and underscore
## Variables are case sensitivity

# Valid variable names
first_name = "Peter"
last_name = "Gakuru"
print("My First name is", first_name, "and Last Name is", last_name)
```

```python
# Invalid Variable names
# 2age=30
# first-name = "Peter"
# @name = "Gakuru"
```

```python
# Case Sensitivity
name = "Peter"
Name = "Gakuru"

name==Name
```

```python
# Understanding Variable Types
# Python is dynamically typed, type of variable is determined at runtime
age=28 #int
height=6.1 #float
name="Peter" #string
is_student=False # bool

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

```python
# Type checking and conversion
type(height)
```

```python
# Converting one data type to another
age=28
print(type(age))

# Type conversion
age_str=str(age)
print(age_str)
print(type(age))
```

```python
# Convert string to int
age='28'
print(type(int(age)))
```

```python
#
height=5.11
type(height)
```

```python
# Converting the float into an integer
int(height)

# Converting float to int
float(int(height))
```

```python
# Dynamic Typing
# Python allows the type of variable to change as the program executes

var=10
print(var, type(var))

var = "Hello"
print(var, type(var))

var=3.14
print(var, type(var))
```

```python
# Input

age=int(input("What is the age: "))
print(age, type(age))
```

#### Creating a Simple Calculator

```python
# Creating a Simple Calculator

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))    

sum = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2

print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)  
```
