### Assignment 6: List Methods



```python
## Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.

import random

random_numbers = [random.randint(1, 20) for _ in range(15)]
print(f"Original List: {random_numbers}")

sorted_numbers = sorted(random_numbers)
print(f"Sorted in ascending order: {sorted_numbers}")

sorted_numbers_desc = sorted(random_numbers, reverse=True)
print(f"Sorted numbers in descending order: {sorted_numbers_desc}")

unique_numbers = list(set(random_numbers))
print(f"List with duplicate removed: {unique_numbers}")
```

### Assignment 8: List of Dictionaries



```python
## Create a list of dictionaries where each dictionary represents a student with keys 'name' and 'score'. Sort the list of dictionaries by the 'score' in descending order and print the sorted list.

students = [
    {'name': 'Gakuru', 'score': 90},
    {'name': 'Gatusso', 'score': 88},
    {'name': 'Kasongo', 'score': 24},
    {'name': 'Les Wanyika', 'score': 80},
    {'name': 'Winning', 'score': 92},
    {'name': 'Malkia', 'score': 90}
]
sorted_students = sorted(students, key=lambda x:x['score'], reverse=True)
print("Sorted students by score in descending order")
for students in sorted_students:
    print(students)
```

### Assignment 9: Matrix Transposition



```python
## Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

def transpose_matrix(matrix):
    transposed = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
    return transposed

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transposed = transpose_matrix(matrix)
print("Original matrix: ")
for row in matrix:
    print(row)
print("Transposed Matrix")
for row in transposed:
    print(row)
```

### Assignment 10: Flattening a Nested List



```python
## Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

def flatten_list(nested_list):   ## Defines a function flatten_list and takes one input nested_list(a list of lists)
    flat_list = [item for sublist in nested_list for item in sublist]  ## for sublist in nested_list → goes through each inner list, e.g. [1, 2, 3], [4, 5, 6],
    return flat_list

nested_list = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

flattened = flatten_list(nested_list)
print("Original Nested List")
print(nested_list)  ## Show the structure before flattening
print("Flattened List")
print(flattened) # Show the result after flattening


# Explanation
## for item in sublist → goes through each number inside that inner list.
## item → each individual number (1, 2, 3, 4, 5, 6, 7, 8, 9).
## All item values are collected into one list called flat_list.
## Meaning: Take all numbers from all inner lists and put them into one single list.
```

### Assignment 14: List Rotation



```python
## Write a function that rotates a list by n positions. Print the original and rotated lists.

def rotate_list(lst, n):
    return lst[n:] + lst[:n]

original_list = [1, 2, 3, 4, 5]
rotated_list = rotate_list(original_list, 2)  ## Calls the function and rotates the list by 2 position
print(f"Original List: {original_list}")
print(f"Rotated List: {rotated_list}")
```

### Assignment 15: List Intersection



```python
## Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

def list_intersection(lst1, lst2):
    return [x for x in lst1 if x in lst2]

# for x in lst1 → go through each item x in lst1.
# if x in lst2 → only keep x if it also exists in lst2.
# Build a list of all items from lst1 that are also in lst2.

lst1 = [1, 2, 3, 4, 5]
lst2 = [3, 4, 5, 6, 7]
intersection = list_intersection(lst1, lst2)
print(f"List 1: {lst1}")
print(f"List 2: {lst2}")
print(f"The Intersection: {intersection}")
```
