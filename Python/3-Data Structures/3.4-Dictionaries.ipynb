#### Introduction to Dictionaries

-  Dictionaries are unordered collection of items.

- They store data in key-value pairs.

- Keys must be unique and immutable (e.g strings, numbers, or tuples), while values can be of any type.

```python
## Creating Dictionaries

empty_dic = {}
print(type(empty_dic))
```

```python
del dict
```

```python
# empty dictionaries

empty_d = dict()
empty_d
```

```python
## Key value pairs

student = {"name":"Kasongo", "age":55,"grade": 40}
print(student)
print(type(student))
```

```python
## Single keys always used - overides the first key
## Check name

student = {"name":"Kasongo", "age":55,"name": 40}
print(student)
```

```python
## Accessing Dictionary Elements
student = {"name":"Kasongo", "age":55,"grade": "A"}
print(student)
```

```python
print(student['grade'])
## Age
print(student['age'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('Last_name'))

## Default value
print(student.get('last_name',"Not Available"))
```

```python
## Modifying Dictionary Elements
## Dictionaries are mutable, so you can add, update or delete elements
## Key must be unique at all times

print(student)
```

```python
## Updating and Adding new key

## Updating
student["age"] = 33
print(student)

## Adding
student["address"] = "Aussie"
print(student)
```

```python
## Deleting Keys

del student['grade']  ## Delete key and Value

print(student)
```

```python
## Dictionary Methods

## All the keys of the dictionary
keys = student.keys()
print(keys)

## Values
values = student.values()
print(values)

## Key value Pairs
items = student.items()
print(items)
```

```python
## SHallow Copy
student_copy = student
print(student_copy)
```

```python
##
student["name"] = "Kasongo1"
print(student)
print(student_copy)
```

```python
## Shallow copy

student_copy1 = student.copy()
```

```python
print(student_copy1)
print(student)
```

```python
student["name"] = "Kasongo2"
print(student_copy1)
print(student)
```

```python
## Iterating Over Dictionaries
## You can use loops to iterate over dictionaries; keys, values, or items

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over values
for values in student.values():
    print(values)
```

```python
## Iterate over key value pairs
for key,value in student.items():
    print(f"{key},{value}")
```

```python
## Nested Dictionaries
## Dictionary inside a dictionary

students = {
    "student1": {"name":"Win", "age":30},
    "student2": {"name":"Parker", "age":42}
}
print(students)
```

```python
## Accessing Nested dictionaries elements

print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
##
students.items()
```

```python
## Iterating over Nested DIctionary

for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comprehension

squares = {x:x**2 for x in range(5)}  # (x:x**2 - squaring each number in the range)
print(squares)

```

```python
## Squares of only even number, using conditions
## Condition dictionary comprehension

even = {x:x**2 for x in range(10) if x % 2==0} 
print(even)
```

```python
## Practical Examples

## Use a dictionary to count the frequency of elements in list

numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1   ## With respect to the key there is an increment
    else:
        frequency[number]=1  ## 
print(frequency)

## Shows the number of times the numbers appear(its frequency)
```

```python
## Merge 2 Dictionaries to one

dict1 = {"a":1, "b":2}
dict2 = {"b":3,"c":4}
merged_dict = {**dict1,**dict2}
print(merged_dict)
```

```python
## Merging

copied = dict1.copy();
copied.update(dict2)
print(copied)
```
