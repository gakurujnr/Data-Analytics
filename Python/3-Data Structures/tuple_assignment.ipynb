### Assignment 1: Creating and Accessing Tuples


```python
## Create a tuple with the first 10 positive integers. Print the tuple.

tpl = tuple(range(1, 11))
print(tpl)
```

### Assignment 2: Accessing Tuple Elements


```python
## Print the first, middle, and last elements of the tuple created in Assignment 1.

print(f"First element: {tpl[0]}")
print(f"Middle element: {tpl[len(tpl) // 2]}")
print(f"Last element: {tpl[-1]}")
```

### Assignment 3: Tuple Slicing


```python
## Print the first three elements, the last three elements, and the elements from index 2 to 5 of the tuple created in Assignment 1.

print(f"First three Elements: {tpl[:3]}")
print(f"Last three Elements: {tpl[-3]}")
print(f"Elements from index 2 to 5: {tpl[1:5]}")
```

### Assignment 4: Nested Tuples


```python
## Create a nested tuple representing a 3x3 matrix and print the matrix. Access and print the element at the second row and third column.

matrix = [
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
]
print("Matrix")
for row in matrix:
    print(row)
print(f"Element at second row and third column: {matrix[1][2]}")
```

### Assignment 5: Tuple Concatenation


```python
### Concatenate two tuples: (1, 2, 3) and (4, 5, 6). Print the resulting tuple.

tpl1 = (1, 2, 3)
tpl2 = (4, 5, 6)
conc = tpl1 + tpl2
print(conc)
```

### Assignment 6: Tuple Methods


```python
## Create a tuple with duplicate elements and count the occurrences of an element. Find the index of the first occurrence of an element in the tuple.

tpl = (1, 2, 2, 3, 4, 4, 4, 5)
print(f"Occurence of 4: {tpl.count(4)}")
print(f"Index of first occurence of 2: {tpl.index(2)}")
```

### Assignment 10: Tuple and List


```python
## Create a tuple with the first 5 positive integers. Convert it to a list, append the number 6, and convert it back to a tuple. Print the resulting tuple.

tpl = (1, 2, 3, 4, 5)
lst = list(tpl)
lst.append(6)
tpl = tuple(lst)
print(tpl)
```

### Assignment 13: Nested Tuple Iteration


```python
## Create a nested tuple and iterate over the elements, printing each element.

nested_tpl = (
    (1,2,3,4), 
    (6,7,8,9),
    (1, "Hello",3.14, "c")
)
for tpl in nested_tpl:
    for elem in tpl:
        print(elem)
```

### Assignment 14: Tuple and Set




```python
## Create a tuple with duplicate elements. Convert it to a set to remove duplicates and print the resulting set.

tpl = (1,2,2,3,3,4, 6,7,7)
uniq = set(tpl)
print(f"Original Tuple: {tpl}")
print(f"Final set: {uniq}")
```

### Assignment 15: Tuple Functions



```python
## Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.


def min_tuple(tpl):
    return min(tpl)

def max_tuple(tpl):
    return max(tpl)

def sum_of_tupl(tpl):
    return sum(tpl)

sample_tpl = (1, 2, 3, 4, 5)
print(f"Minimum tuple: {min_tuple(sample_tpl)}")
print(f"Maximum tuple: {max_tuple(sample_tpl)}")
print(f"Sum of tuple: {sum_of_tupl(sample_tpl)}")
```
