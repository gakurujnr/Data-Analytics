### Assignment 11: Dictionary Filtering

Create a dictionary with the first 10 positive integers as keys and their squares as values. Create a new dictionary containing only the key-value pairs where the key is even. Print the new dictionary.

```python
## Even

d1 = {x: x**2 for x in range(1, 11)}
even = {k: v for k, v in d1.items() if k % 2 == 0}
print(even)
```

### Assignment 12: Dictionary Key and Value Transformation

```python
## Swapping
d = {x: x**2 for x in range(1, 6)}
swap = {v: k for k, v in d.items()}
print(swap)
```

### Assignment 13: Default Dictionary

Create a default dictionary where each key has a default value of an empty list. Add some elements to the lists and print the dictionary.

```python
## Value of empty list

from collections import defaultdict 
## It gives you a dictionary that automatically creates a default value for keys that don't exist yet — 
# so you don't get a KeyError when you access or update a missing key. 


dfl = defaultdict(list)
dfl['a'].append(1)
dfl['a'].append(2)
dfl['b'].append(3)
print(dfl)
```

```python
from collections import defaultdict

default_dict = defaultdict(list)
default_dict['a'].append(1)
default_dict['a'].append(2)
default_dict['b'].append(3)
print(default_dict)
```

### Assignment 14: Counting with Dictionaries

Write a function that takes a string and returns a dictionary with the count of each character in the string. Print the dictionary.

```python
def count_chars(s):
    count_dict = {}
    for char in s:
        count_dict[char] = count_dict.get(char, 0) + 1
    return count_dict

string = "hello world"
print(count_chars(string))
```

### Assignment 15: Dictionary and JSON

Create a dictionary representing a book with keys 'title', 'author', 'year', and 'genre'. Convert the dictionary to a JSON string and print it.

```python
import json   ## Brings in Python’s built-in json module so you can convert between Python objects and JSON text.

##  turns a Python dictionary (book) into a JSON-formatted string (book_json) and prints that string — 
# json.dumps performs the conversion (serialization).

book = {
    'title': 'TAo Kill a Mockingbird',
    'author': 'Harper Lee',
    'year': 1960,
    'genre': 'Fiction'
}
book_json = json.dumps(book)  ## Performs the conversion (serialization) of the Python dictionary (book) into a JSON-formatted string (book_json).
print(book_json)
```
