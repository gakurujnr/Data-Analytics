#### Introduction to Tuples

Tuples are ordered collection of items that are immutable. They are similar to lists, but their immutability makes them different.

```python
## Creating Tuple

empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
## Creating empty list and tuple

lst = list()
print(type(lst))
tpl = tuple()
print(type(tpl))
```

```python
## Create tuple

numbers = tuple([1, 2, 3, 4, 5])
numbers
```

```python
## 
list((1,2,3,4,5))
```

```python
## Mixed Tuple

mixed_tuple = (1, "Hello World", 3.142, True)
print(mixed_tuple)
```

```python
## Accessing tuple elements

numbers 
# numbers[2]
print(numbers[-1])
```

```python
numbers[0:4]
```

```python
## Tuple operation - Concatination, Repetition and slicing

concatination_tuple = numbers + mixed_tuple
print(concatination_tuple)
```

```python
##

mixed_tuple * 3
```

```python
## Immutable Nature of Tuples
# Tuples are immutable, meaning their elements cannot be changed once assigned

lst = [1,2,3,4,5]
print(lst)

lst[1] = "Win"
print(lst)
```

```python
## Once assigned can't be changed

numbers[1] = "Win"
```

```python
## Tuple Methods

print(numbers.count(1))
print(numbers.index(3))
```

```python
## Packing and Unpacking Tuples
# Packing
packed_tuple = 1, "Hello", 3.14
print(packed_tuple)
```

```python
## Unpacking a Tuple
a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

```python
# Unpacking with Star

numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle)  #* Picks all the elements from where you are starting to the last element
print(last)
```

```python
## Example for Nested Tuple

lst = [[1,2,3,4], [6,7,8,9],[1, "Hello",3.14, "c"]]
print(lst)

lst[0][0:3]   # Slicing in a nested list
```

```python
##

lst = [[1,2,3,4], [6,7,8,9],(1, "Hello",3.14, "c")]
print(lst)

lst[2][0:3] 
```

```python
## Nested Tuple

nested_tuple = ((1,2,3), ("a", "b", "c"), (True, False))

## Access the elements inside a tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over nested tuples

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
    print()
```

### Concludion

Tuples are versatile and useful in many real-world scenarios where an immutable and ordered collection of items is required. 

They are commonly used in data structures, function arguments and return values, and as dictionary keys. 

Understanding how to leverage tuples effectively can improve the efficiency and readability of your Python code.
