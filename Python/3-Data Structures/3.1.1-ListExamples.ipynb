### Real-World Examples Using Lists in Python

- Lists are one of the most commonly used data structures in Python, are versatile and easy to use.



#### Example 1: Manage A To Do List

```python
## Create a To Do List To Keep Track of Tasks

to_do_list = ["Buy Groceries", "Clean the House", "Finish Homework", "Call Mom"]

## Adding Tasks
to_do_list.append("Pay Bills")
to_do_list.append("Exercise")

## Removing a completed task
to_do_list.remove("Clean the House")

## Check if a task is in the list
if "Finish Homework" in to_do_list:
    print("Don't forget to pay the utility bills")

print("To Do List Remaining")
for task in to_do_list:
    print(f"-{task}")
```

#### Organizing Students Grades


- Create a list to store and calculate average grades for students

```python
## Organizing grades (List)
grades = [85, 92, 78, 96, 88]
print(f"Grades are: {grades}")

## Adding new grades
grades.append(95)
grades.append(74)

# Calculating the average grade
average_grade = sum(grades) / len(grades)
print(f"Average grade: {average_grade:.2f}")

## Finding the highest and the lowest grades
highest_grade = max(grades)
lowest_grades = min(grades)
print(f"Highest Grade: {highest_grade}")
print(f"Lowest Grade: {lowest_grades}")
```

#### Use a list to manage Inventory items in a store

```python
## Managing Inventory
inventory = ["apple", "banana", "orange", "grape"]
print(f"These are the items in the inventory: {inventory}")

## Adding a ne=w item
inventory.append("kiwi")
inventory.append("mango")

## Removing an item
inventory.remove("banana")

## Checking if an item is in stock
item = "oranges"
if item in inventory:
    print(f"{item} are in stock.")
else:
    print(f"{item} are out of stock.")
```

#### Use a list to collect and analyze user feedback

```python
# Collecting User Feedback
feedback = ["Great Service", "Very Satisfied", "Could do better", "Needs Improvement"]
print(f"These are the feedbacks displayed: {feedback}")

## Adding feedback
feedback.append("Excellent Experience")
feedback.append("Not Happy with the service")

## Counting Specific Feedback
positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())
print(f"Positive Feedback Count: {positive_feedback_count}")

## Print all feedback
print("User Feedback")
for comment in feedback:
    print(f"- {comment}")

```
