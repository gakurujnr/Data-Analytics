## Sets

Sets are a built‑in data type in Python used to store collections of unique items. 

They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. 

Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

```python
## Create a set
# Ste uses curly brackets
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
## Empty set

my_empty_set = set()
print(type(my_empty_set))
```

```python
##

my_empty_set = set({1,2,3,4,5})
print(my_empty_set)
```

```python
##

my_set = set({1,2,3,4,5,6,5,4})
print(my_set)

## Sets do not allow duplicate element
```

```python
## Basic set operations
# Adding and removing elements

my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
## Removing elements from a set

my_set.remove(6)
print(my_set)
```

```python
## Remove element thats not present
my_set.remove(10)
print(my_set)
```

```python
## Discardinf
my_set.discard(10)
print(my_set)
```

```python
## Removing Random element using 'Pop Method'
# Pop - removes first element
remove_element = my_set.pop()
print(remove_element)
print(my_set)
```

```python
## Clear all elements

my_set.clear()
print(my_set)
```

```python
## Set membership test
## Whether an element is a member of the set or not

my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)

## If the element is present in the set the output is True else its false
```

#### Mathematical Operation

```python
## Mathematical Operation

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Union
## Combining the sets if you have two sets

union_set = set1.union(set2)
print(union_set)

## Intersection
# It will give us a common element in both set
intersection_set = set1.intersection(set2)
print(intersection_set)

# Intersection_update
## It keeps ONLY the elements that appear in BOTH sets.
# Everything else in set1 is removed.
set1.intersection_update(set2)
print(set1)
```

```python
## Difference
# Remove common elements from set2

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

Difference_set = set1.difference(set2)
print(Difference_set)


```

```python
## Difference
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

set2.difference(set1)
```

```python
## Symmetric Difference
## Removing common elements from both sets
set1.symmetric_difference(set2)
```

```python
## Stes Methods

set1 = {1,2,3,4,5}
set2 = {3,4,5}

## is subset
## Means all elements present in set1 are they available in set2
print(set1.issubset(set2))

## Is Superset
## Check if set1 has all the elements of set2
print(set1.issuperset(set2))
```

```python
#### Practical Example
# Counting Unique words in text

text = "In this tutorial we are discussing about sets"
words = text.split()
# print(words)

## Convert List of words to set to get unique words
unique_words = set(words)
print(unique_words)

## Print the Length of the unique words
print(f"The Length: {len(unique_words)}")
```
